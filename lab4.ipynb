```python
from google.colab import drive
import sys
import cv2
import numpy as np
%matplotlib inline
import matplotlib.pyplot as plt
drive.mount('/content/drive')
sys.path.append('/content/drive/My Drive')
image = cv2.imread('/content/drive/My Drive/lab4image.jpg')
rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.title("Исходное изображение")
plt.imshow(rgb_image)
```

```python
plt.hist(rgb_image.ravel(), bins=256, range=[0,256])
plt.title('Histogram')
plt.show()
```

Пороговая фильтрация

```python

_, thresh = cv2.threshold(rgb_image, 120 , 255, cv2.THRESH_BINARY)

plt.imshow(thresh, cmap='gray')
plt.show()
```

```python
threshold = 120
image = rgb_image

ret, thresh1 = cv2.threshold(image, threshold, 255, cv2.THRESH_BINARY)
ret, thresh2 = cv2.threshold(image, threshold, 255, cv2.THRESH_BINARY_INV)
ret, thresh3 = cv2.threshold(image, threshold, 255, cv2.THRESH_TRUNC)
ret, thresh4 = cv2.threshold(image, threshold, 255, cv2.THRESH_TOZERO)
ret, thresh5 = cv2.threshold(image, threshold, 255, cv2.THRESH_TOZERO_INV)
titles = ['Grayscale Image', 'BINARY', 'BINARY_INV', 'TRUNC', 'TOZERO', 'TOZERO_INV']
images = [image, thresh1, thresh2, thresh3, thresh4, thresh5]
plt.figure(figsize=(15, 8))
for i in np.arange(len(images)):
    plt.subplot(2, 3, i + 1)
    plt.imshow(images[i], 'gray', vmin=0, vmax=255)
    plt.title(titles[i])
    plt.xticks([])
    plt.yticks([])
plt.show()
```

 Бинаризация Оцу

```python

image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
ret1, thresh1 = cv2.threshold(image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
thresh2 = cv2.adaptiveThreshold(image, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 15, 5)
thresh3 = cv2.adaptiveThreshold(image, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 15, 5)
titles = ['Original Image', "Otsu's Thresholding (Otsu's threshold = %d)" % ret1,
          'Adaptive Mean Thresholding', 'Adaptive Gaussian Thresholding']
images = [image, thresh1, thresh2, thresh3]

plt.figure(figsize=(10, 8))
for i in np.arange(len(images)):
    plt.subplot(2, 2, i + 1)
    plt.imshow(images[i], 'gray')
    plt.title(titles[i])
    plt.xticks([])
    plt.yticks([])
plt.show()
```
